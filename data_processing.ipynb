```python
outside_data_root = './data/outside_concept/'
```

```python
food_concepts = pd.read_csv(outside_data_root + 'CLAIMS_7_years_HUMAN FOOD_data.csv')
us_food_concepts = pd.read_excel(outside_data_root + 'us_food_for_synthetic.xlsx')
```

```python
us_food_concepts.head()
```

```python

mask = (~food_concepts['ConceptText'].isna()) & (food_concepts['Country']=='United States')
food_concepts = food_concepts.loc[mask]
```

```python
food_concepts.drop(['CE_MostRelevant', 'CE_LeastRelevant',
       'CE_EspeciallyDifficult', 'CE_NewAndDifferent', 'CE_WordCode'], axis=1, inplace=True)
food_concepts['ConceptText'] = food_concepts['ConceptText'].apply(lambda x: hp.extract_concept_info(x))


food_concepts['CAT2'] = food_concepts['CAT2'].apply(lambda x: hp.clean_category_name(x) if not pd.isna(x) else x)
food_concepts['CAT3'] = food_concepts['CAT3'].apply(lambda x: hp.clean_category_name(x) if not pd.isna(x) else x)
```

```python
food_concepts['CAT3'].isna().sum()
```

```python
# save the cleaned data 
food_concepts.to_excel(outside_data_root + 'cleaned_food_concepts.xlsx', index=False)
```
